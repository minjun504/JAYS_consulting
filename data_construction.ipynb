```python
import pandas as pd
```

```python
beach_data = "data/beach_monthly.csv"
incident_data = "data/incidents.csv"
pricing_data = "data/pricing.csv"

beach_df = pd.read_csv(beach_data)
incident_df = pd.read_csv(incident_data)
pricing_df = pd.read_csv(pricing_data)
```

```python
incident_df["dte_occurred"] = pd.to_datetime(incident_df["dte_occurred"], format="%d/%m/%Y")
incident_df["month"] = incident_df["dte_occurred"].dt.strftime("%m/%Y")
incident_df["month"]
```

```python
aggregated_df = incident_df.groupby(["month", "beach_nm"]).agg(
    n_incidents=("incident_id", "nunique"),  
    n_people=("incident_id", "size"),        
    n_mild=("inc_type", lambda x: (x == "Mild").sum()),
    n_closecall=("inc_type", lambda x: (x == "CloseCall").sum()),
    n_severe=("inc_type", lambda x: (x == "Severe").sum())
).reset_index()

print(aggregated_df)
```

# inc_cause, surfer, has null values

```python
agg = incident_df.groupby(["beach_nm", "month"]).agg(
    n_incidents=("incident_id", "nunique"),
    n_people=("incident_id", "size"),
)
types = (pd.crosstab([incident_df["beach_nm"], incident_df["month"]], incident_df["inc_type"])
           .rename(columns=lambda c: f"n_{c.lower()}"))
agg = agg.join(types).reset_index()

panel = beach_df.merge(agg, on=["beach_nm", "month"], how="left")
cols = ["n_incidents", "n_people", "n_mild", "n_closecall", "n_severe"]
panel[cols] = panel[cols].fillna(0).astype(int)
```

```python
panel
```

```python

bm = beach_df.copy()
bm["month_key"] = pd.to_datetime(bm["month"], format="%m/%Y").dt.to_period("M")   

inc = incident_df.copy()
for c in ["dte_occurred", "dte_reported"]:
    inc[c] = pd.to_datetime(inc[c], format="%d/%m/%Y")
inc["month_key"] = inc["dte_occurred"].dt.to_period("M")   

person = inc.merge(
    bm.drop(columns="month"),
    on=["beach_nm", "month_key"],
    how="left",
    validate="many_to_one",
    indicator=True,
)
assert len(person) == len(incident_df) == 33_941
assert (person["_merge"] == "both").all()
person = person.drop(columns="_merge")

person["incident_size"]   = person.groupby("incident_id")["person_no"].transform("size")
person["on_duty"]         = (person["n_lifeguards"] > 0).astype(int)
person["surfer"]    = person["surfer"].astype("string").fillna("unknown")
person["inc_cause"] = person["inc_cause"].fillna("unknown")      
person["inc_type"]  = pd.Categorical(person["inc_type"],
                                     categories=["Mild", "CloseCall", "Severe"], ordered=True)
```
